## A.1 / Описательная аналитика: что и как?

> **Описательная аналитика (Descriptive Analytics)** — направление анализа данных, целью которого является количественная характеристика наблюдаемых явлений и процессов: их состояния, структуры, динамики, вариативности и взаимосвязей. Она позволяет бизнесу держать руку на пульсе и следить за самыми важными метриками. Фундаментальные вопросы: что происходит и как это происходит (в том числе когда).

В клиентской аналитике она позволяет систематизировать знания о клиентской базе, потребительском поведении и экономических результатах, выявлять изменения и закономерности, формировать гипотезы для последующих исследований. Описание может выступать самостоятельной аналитической задачей, а не только подготовка к более сложному исследованию.

### Мониторинг метрик как система показателей и сигналов

Мониторинг связывает бизнес-результат с наблюдаемыми компонентами клиентского пути. Сигнал — изменение показателя относительно сопоставимого периода, исторического диапазона или согласованного порога; сам по себе он не объясняет причину изменения.

  <td>Рост CAC при неизменном LTV</td>
</tr>
</tbody>
</table>


Мониторинг метрик, как правило, автоматизируется с помощью системы BI-отчетности. Это позволяет регулярно отслеживать состояние бизнеса, динамику ключевых показателей и своевременно замечать отклонения.

Если аналитик обнаруживает нетипичное изменение метрики, он формулирует гипотезы о его возможных причинах и проводит серию небольших исследований. Например, анализирует динамику отдельных клиентских сегментов, декомпозирует метрику на составляющие или сравнивает поведение разных когорт.

Цель таких исследований — понять, за счет чего изменилась метрика и на какие факторы бизнесу стоит обратить внимание.

```python
def generate_data(path=DATA, seed=42, n=6000):
    rng = np.random.default_rng(seed)
    ids = np.arange(1, n + 1)
    signup = START + pd.to_timedelta(rng.integers(0, 365, n), unit='D')
    cities = rng.choice(['Москва', 'Санкт-Петербург', 'Казань'], n, p=[.55,.30,.15])
    channels = rng.choice(['organic','paid_search','referral','social'], n, p=[.38,.28,.19,.15])
    preferred = rng.choice(['iOS','Android','Web'], n, p=[.42,.45,.13])
    profile = rng.choice(3, n, p=[.32,.48,.20])
    rates = np.array([.65, 2.7, 6.0])[profile] * rng.lognormal(0,.30,n)
    budget = rng.lognormal(np.log(1050),.35,n)
    decay = np.array([.27,.10,.035])[profile]
    cust = pd.DataFrame({'customer_id':ids,'signup_date':signup,'city':cities,'acquisition_channel':channels})
    sessions, orders = [], []
    sid = oid = 0
    for i in range(n):
        for m in MONTHS[MONTHS >= signup[i].to_period('M')]:
            lo = max(m.start_time, signup[i]); hi = min(m.end_time.normalize()+pd.Timedelta(days=1), END)
            days = (hi-lo).days
            age = m.ordinal - signup[i].to_period('M').ordinal
            seasonal = 1 + (.30 if m.month in (11,12) else -.18 if m.month in (6,7) else 0)
            lam = rates[i] * np.exp(-decay[i]*age) * seasonal * days / m.days_in_month
            count = rng.poisson(lam) + int(age == 0)  # registration-period visit
            for _ in range(count):
                sid += 1
                t = lo + pd.Timedelta(days=int(rng.integers(days)), minutes=int(rng.integers(8*60, 22*60)))
                platform = preferred[i] if rng.random()<.9 else rng.choice(['iOS','Android','Web'])
                view = t+pd.Timedelta(minutes=1) if rng.random()<.91 else pd.NaT
                cart = view+pd.Timedelta(minutes=2) if pd.notna(view) and rng.random()<(.72 if platform!='Web' else .58) else pd.NaT
                checkout = cart+pd.Timedelta(minutes=3) if pd.notna(cart) and rng.random()<.79 else pd.NaT
                p = [.67,.81,.89][profile[i]] - (.10 if platform=='Web' else 0)
                purchase = checkout+pd.Timedelta(minutes=2) if pd.notna(checkout) and rng.random()<p else pd.NaT
                sessions.append((sid,ids[i],platform,t,view,cart,checkout,purchase))
                if pd.isna(purchase): continue
                oid += 1
                promo = rng.random() < (.37 if age<2 else .23)
                gmv = round(max(250, budget[i]*rng.lognormal(0,.35)),2)
                discount = round(gmv*rng.uniform(.08,.25),2) if promo else 0.
                revenue = round(gmv-discount,2)
                variable_cost = round(gmv*rng.uniform(.66,.79)+rng.uniform(55,120),2)
                margin = round(revenue-variable_cost,2)
                promised = 35 if cities[i]=='Москва' else 40
                late = rng.random() < (.14 + (.09 if m.month in (11,12) else 0) + (.04 if cities[i]=='Москва' else 0))
                actual = int(promised+rng.integers(5,35)) if late else int(rng.integers(15,promised+1))
                support = rng.random()<(.26 if late else .035)
                orders.append((oid,sid,ids[i],purchase,gmv,discount,revenue,variable_cost,margin,int(promo),promised,actual,int(support)))
    ses = pd.DataFrame(sessions, columns=['session_id','customer_id','platform','session_start','product_view','add_to_cart','checkout','purchase'])
    ords = pd.DataFrame(orders, columns=['order_id','session_id','customer_id','order_ts','gmv','discount','net_revenue','variable_cost','contribution_margin','promo_used','promised_minutes','actual_minutes','support_contact'])
    cust.to_csv(path/'customers.csv',index=False)
    ses.sort_values('session_start').to_csv(path/'sessions.csv',index=False)
    ords.sort_values('order_ts').to_csv(path/'orders.csv',index=False)

if not all((DATA/f'{name}.csv').exists() for name in ['customers','sessions','orders']):
    generate_data()
customers = pd.read_csv(DATA/'customers.csv', parse_dates=['signup_date'])
steps = ['session_start','product_view','add_to_cart','checkout','purchase']
sessions = pd.read_csv(DATA/'sessions.csv', parse_dates=steps)
orders = pd.read_csv(DATA/'orders.csv', parse_dates=['order_ts'])
```

```python
# Единый расчетный слой: функции не изменяют входные таблицы.
from dataclasses import dataclass
from textwrap import fill
from matplotlib.ticker import FuncFormatter, PercentFormatter
from matplotlib.patches import Patch
from matplotlib.lines import Line2D
from io import BytesIO
from IPython.display import Image, HTML, display
import base64, json
from html import escape
from functools import lru_cache

MONTHS = pd.PeriodIndex(MONTHS, freq='M')
MONTHS = MONTHS[MONTHS.end_time.normalize() <= AS_OF.normalize()]
assert len(MONTHS) and np.all(np.diff(MONTHS.asi8)==1)
assert START.normalize() == MONTHS[0].start_time
REPORT_END = MONTHS[-1].end_time.normalize()
DATES = pd.date_range(START, REPORT_END)
AGES = np.arange(len(MONTHS))
CHURN_DAYS, DEAD_MONTHS, LOOKBACK_DAYS, FORWARD_DAYS, CONVERSION_DAYS = 60, 3, 180, 60, 30
plt.rcParams['figure.autolayout'] = False
LABELS.update({'sessions':'Сессии','mean_DAU':'DAU — средняя дневная аудитория','DAU_MAU':'DAU/MAU — регулярность','wau':'WAU — недельная аудитория','payer_share':'Доля покупателей среди MAU','margin_rate':'Маржинальность','Active':'Active — активные','At Risk':'At Risk — под риском','Churn':'Churn — отток','Dead':'Dead — глубокая неактивность','New':'New — новички','revenue_ltv':'LTV(R) — накопленная выручка','margin_ltv':'LTV(M) — накопленная маржа','ltv_cac':'LTV(M)/CAC','ltv_r_cac':'LTV(R)/CAC','net_value':'Маржа за вычетом CAC','registered_30':'Регистрация','visited_30':'Первая сессия','paid_30':'Первая покупка','churned_customers':'Клиенты 60+ дней без заказов','churn_rate':'Доля невозврата','quick_ratio':'Quick Ratio — приток / потери','rate':'Доля повторных покупателей','order_rate':'Заказы на исходного покупателя','margin_per_original':'Маржа на исходного покупателя','survival':'Выживаемость','first_buyers':'Новые покупатели'})
STATE3 = ['Active','At Risk','Churn']
STATE4 = ['New','Active','Churn','Dead']
COL3 = [PALETTE['brown'],PALETTE['ochre'],PALETTE['burgundy']]
COL4 = [PALETTE['ochre'],PALETTE['brown'],PALETTE['burgundy'],PALETTE['gray']]

def div(a,b):
    return a / b.replace(0,np.nan) if hasattr(b,'replace') else a/b if b else np.nan

def num(x, decimals=0):
    return '—' if pd.isna(x) else f'{x:,.{decimals}f}'.replace(',',' ').replace('.',',')
def pct(x): return '—' if pd.isna(x) else f'{x:.1%}'.replace('.',',')
def rub(x): return num(x)+' ₽' if pd.notna(x) else '—'
def label(x): return LABELS.get(x,str(x))
def complete_months(days): return MONTHS[(MONTHS+1).to_timestamp()+pd.Timedelta(days=days)<=REPORT_END+pd.Timedelta(days=1)]
def by_month(frame, date, key=None):
    g=frame.groupby(frame[date].dt.to_period('M'))
    return (g.size() if key is None else g[key].nunique()).reindex(MONTHS,fill_value=0)
def cohort_mask(cohorts, ages=AGES, latest=None):
    latest = cohorts if latest is None else latest
    return pd.DataFrame(latest.asi8[:,None]+np.asarray(ages)[None,:]<=MONTHS[-1].ordinal,index=cohorts,columns=ages)
def age_matrix(frame, metric, index, age='age'):
    return frame.groupby(['cohort',age])[metric].sum().unstack(fill_value=0).reindex(index=index,columns=AGES,fill_value=0).fillna(0)
def weighted_mean(matrix, sizes):
    return div(matrix.mul(sizes,axis=0).sum(min_count=1),matrix.notna().mul(sizes,axis=0).sum())

def funnel(frame):
    valid=frame[steps[0]].notna(); totals=[int(valid.sum())]
    for prev,curr in zip(steps,steps[1:]):
        valid &= frame[curr].notna() & frame[curr].ge(frame[prev]); totals.append(int(valid.sum()))
    out=pd.DataFrame({'sessions':totals},index=steps)
    out['overall_cr']=div(out.sessions,out.sessions.iloc[0]); out['step_cr']=div(out.sessions,out.sessions.shift()); out['lost_sessions']=out.sessions.shift()-out.sessions
    assert out.sessions.is_monotonic_decreasing
    return out

def base_dynamics(events,time_col):
    active=events.groupby(events[time_col].dt.to_period('M')).customer_id.agg(set)
    prev,hist=set(),set(); rows=[]
    for m in MONTHS:
        cur=active.get(m,set()); new=cur-hist; retained=cur&prev; react=(cur&hist)-prev; churn=prev-cur
        rows.append((m,len(prev),len(new),len(retained),len(react),len(churn),len(cur),len(hist-prev)))
        prev=cur; hist|=cur
    d=pd.DataFrame(rows,columns=['month','opening','new','retained','reactivated','churned','closing','dormant']).set_index('month')
    d['net_growth']=d.new+d.reactivated-d.churned; d['quick_ratio']=div(d.new+d.reactivated,d.churned)
    assert d.closing.eq(d.opening+d.net_growth).all()
    return d

def km_table(data):
    ages=np.arange(int(data.duration.max())+1)
    risk=np.array([data.duration.ge(t).sum() for t in ages]); events=np.array([(data.duration.eq(t)&data.event).sum() for t in ages]); censored=np.array([(data.duration.eq(t)&~data.event).sum() for t in ages])
    survival=np.cumprod(1-events/risk)
    assert np.array_equal(risk[1:],(risk-events-censored)[:-1]) and events.sum()==data.event.sum() and censored.sum()==(~data.event).sum()
    out=pd.DataFrame({'at_risk':risk,'events':events,'censored':censored,'survival':survival},index=ages)
    assert out.survival.is_monotonic_decreasing and out.survival.between(0,1).all()
    return out

# Вычисляем ежедневную историю последнего заказа одним проходом (без повторных groupby на каждую дату).
def purchase_history(o, first_order):
    ids=first_order.index
    days=o.assign(day=o.order_ts.dt.normalize()).drop_duplicates(['customer_id','day'])
    purchase=np.zeros((len(ids),len(DATES)),dtype=bool); ix=ids.get_indexer(days.customer_id); dx=DATES.get_indexer(days.day); valid=(ix>=0)&(dx>=0); purchase[ix[valid],dx[valid]]=True
    day_index=np.arange(len(DATES)); last_idx=np.maximum.accumulate(np.where(purchase,day_index,-1),axis=1)
    known=last_idx>=0; recency=np.where(known,day_index-last_idx,-1)
    churn=recency>=CHURN_DAYS
    daily=pd.DataFrame({'customer_base':known.sum(axis=0),'churned_customers':churn.sum(axis=0)},index=DATES)
    daily['inactive_rate']=div(daily.churned_customers,daily.customer_base)
    rolling=[]
    for k in range(LOOKBACK_DAYS-1,len(DATES)-FORWARD_DAYS):
        base=purchase[:,k-LOOKBACK_DAYS+1:k+1].any(axis=1); returned=purchase[:,k+1:k+FORWARD_DAYS+1].any(axis=1)&base
        rolling.append((DATES[k],int(base.sum()),int(returned.sum()),int((base&~returned).sum())))
    rolling=pd.DataFrame(rolling,columns=['date','customer_base','returned_customers','churned_customers']).set_index('date')
    rolling['churn_rate']=div(rolling.churned_customers,rolling.customer_base)
    assert (rolling.returned_customers+rolling.churned_customers==rolling.customer_base).all()
    idx=DATES.get_indexer(MONTHS.end_time.normalize()); month_start=DATES.get_indexer(MONTHS.start_time)
    month_buy=np.column_stack([purchase[:,a:b+1].any(axis=1) for a,b in zip(month_start,idx)])
    return daily,rolling,month_buy,recency[:,idx],known[:,idx]

def first_payback(ratio):
    """Первый наблюдаемый календарный возраст с LTV/CAC >= 1."""
    reached = ratio.dropna().loc[lambda x: x.ge(1)]
    return int(reached.index[0]) if not reached.empty else np.nan

def payback_label(age):
    return 'Не достигнута на наблюдаемом горизонте' if pd.isna(age) else f'M{int(age)} ({int(age)+1}-й месяц)'

def calculate_scope(channel=None):
    # Канал — атрибуция при привлечении, а не платформа текущей сессии.
    c=customers if channel is None else customers[customers.acquisition_channel.eq(channel)]
    ids=c.customer_id
    ss=sessions[sessions.customer_id.isin(ids)].copy(); o=orders[orders.customer_id.isin(ids)].copy()
    a=acq if channel is None else acq[acq.acquisition_channel.eq(channel)]
    sp=spend if channel is None else spend[spend.acquisition_channel.eq(channel)]
    ss=ss[ss.session_start.dt.normalize().le(REPORT_END)]; o=o[o.order_ts.dt.normalize().le(REPORT_END)]
    fo=o.groupby('customer_id').order_ts.min().rename('first_order'); lo=o.groupby('customer_id').order_ts.max().rename('last_order')
    coh=fo.dt.to_period('M').rename('cohort'); sizes=coh.value_counts().reindex(MONTHS,fill_value=0).sort_index()
    m=o.groupby('month').agg(active_customers=('customer_id','nunique'),orders=('order_id','size'),revenue=('net_revenue','sum'),margin=('contribution_margin','sum')).reindex(MONTHS,fill_value=0)
    m['frequency']=div(m.orders,m.active_customers); m['aov']=div(m.revenue,m.orders); m['arppu']=div(m.revenue,m.active_customers); m['margin_rate']=div(m.margin,m.revenue)
    m['gmv']=o.groupby('month').gmv.sum().reindex(MONTHS,fill_value=0); m['promo_cost_per_buyer']=div(o.groupby('month').discount.sum().reindex(MONTHS,fill_value=0),m.active_customers); m['cost_to_serve']=div(o.groupby('month').variable_cost.sum().reindex(MONTHS,fill_value=0),m.orders)
    assert np.allclose(m.revenue,(m.active_customers*m.frequency*m.aov).fillna(0))
    am=pd.DataFrame(index=MONTHS)
    for typ,col in [('app_install','installs'),('web_visit','web_entries')]: am[col]=a[a.entry_type.eq(typ)].groupby('entry_month').size().reindex(MONTHS,fill_value=0)
    am['registrations']=by_month(c,'signup_date'); am['registered_base']=am.registrations.cumsum(); am['sessions']=by_month(ss,'session_start'); am['mau']=by_month(ss,'session_start','customer_id'); am['mpu']=m.active_customers; am['first_buyers']=coh.value_counts().reindex(MONTHS,fill_value=0)
    am['arpu']=div(m.revenue,am.mau); am['arppu']=m.arppu; am['payer_share']=div(am.mpu,am.mau); assert am.mpu.le(am.mau).all()
    days=ss.assign(day=ss.session_start.dt.normalize())[['customer_id','day']].drop_duplicates(); dau=days.groupby('day').size().reindex(DATES,fill_value=0)
    eng=pd.DataFrame({'mean_DAU':dau.groupby(dau.index.to_period('M')).mean(),'MAU':am.mau}); eng['DAU_MAU']=div(eng.mean_DAU,eng.MAU); eng['active_days_per_MAU']=div(days.groupby(days.day.dt.to_period('M')).size().reindex(MONTHS,fill_value=0),eng.MAU); eng['sessions_per_MAU']=div(am.sessions,eng.MAU)
    assert np.allclose(eng.DAU_MAU,eng.active_days_per_MAU/eng.index.days_in_month,equal_nan=True)
    weeks=pd.period_range(START,REPORT_END,freq='W-SUN'); weeks=weeks[(weeks.start_time>=START)&(weeks.end_time.normalize()<=REPORT_END)]
    wau=ss.groupby(ss.session_start.dt.to_period('W-SUN')).customer_id.nunique().reindex(weeks,fill_value=0).rename('wau')
    f=funnel(ss); platform_cr=pd.DataFrame({p:funnel(g).overall_cr for p,g in ss.groupby('platform')})
    a30=a[a.entry_month.isin(complete_months(30))]; s30=sp[sp.entry_month.isin(complete_months(30))]; app=a30[a30.entry_type.eq('app_install')]
    entry=pd.Series({'Install':len(app),'Registration':app.registered_30.sum(),'First session':app.visited_30.sum(),'First order':app.paid_30.sum()}); assert entry.is_monotonic_decreasing
    rank=o.sort_values(['customer_id','order_ts','order_id']); gaps=rank.groupby('customer_id').order_ts.diff().dt.total_seconds().dropna()/86400
    second=rank[rank.groupby('customer_id').cumcount().eq(1)].set_index('customer_id').order_ts
    mature=fo[fo+pd.Timedelta(days=90)<=REPORT_END+pd.Timedelta(days=1)]; rep=pd.DataFrame({'eligible':len(mature),'repeated':[int(second.reindex(mature.index).lt(mature+pd.Timedelta(days=h)).sum()) for h in [7,14,30,60,90]]},index=[7,14,30,60,90]); rep['rate']=div(rep.repeated,rep.eligible)
    window=MONTHS[-6:]; base_ids=c.loc[c.signup_date.lt(window[0].start_time),'customer_id']; behavior=o[o.month.isin(window)&o.customer_id.isin(base_ids)].groupby('customer_id').agg(orders=('order_id','size'),revenue=('net_revenue','sum'),margin=('contribution_margin','sum'),promo_orders=('promo_used','sum'),discount=('discount','sum')).reindex(base_ids,fill_value=0); behavior['aov']=div(behavior.revenue,behavior.orders); behavior['promo_share']=div(behavior.promo_orders,behavior.orders)
    delay=(fo-c.set_index('customer_id').signup_date.reindex(fo.index)).dt.total_seconds()/86400
    daily,rolling,bought,recency,known=purchase_history(o,fo)
    # Единый расчет трех состояний по 60-дневному определению.
    counts={k:pd.DataFrame(np.nan,index=MONTHS,columns=AGES) for k in STATE3}
    states=pd.DataFrame(index=fo.index,columns=MONTHS,dtype=object); missed=np.zeros(len(fo),dtype=int)
    for k,month in enumerate(MONTHS):
        active=bought[:,k]; churn=recency[:,k]>=CHURN_DAYS; at_risk=known[:,k]&~active&~churn
        assert np.array_equal(active.astype(int)+at_risk.astype(int)+churn.astype(int),known[:,k].astype(int))
        for name,mask in zip(STATE3,[active,at_risk,churn]):
            for co,v in pd.Series(mask,index=fo.index).groupby(coh).sum().items():
                if co<=month: counts[name].loc[co,month.ordinal-co.ordinal]=v
        missed=np.where(active,0,missed+1); new=coh.eq(month).to_numpy()
        states.loc[new,month]='New'; states.loc[known[:,k]&~new&active,month]='Active'; states.loc[known[:,k]&~active&(missed<DEAD_MONTHS),month]='Churn'; states.loc[known[:,k]&~active&(missed>=DEAD_MONTHS),month]='Dead'
    rates={k:v.div(sizes.replace(0,np.nan),axis=0) for k,v in counts.items()}; structure=pd.DataFrame({k:weighted_mean(v,sizes) for k,v in rates.items()}); assert np.allclose(structure.dropna().sum(axis=1),1)
    assert np.allclose(sum(rates.values()).stack().dropna(),1)
    mf=base_dynamics(ss,'session_start'); pf=base_dynamics(o,'order_ts'); assert mf.closing.eq(am.mau).all() and pf.closing.eq(am.mpu).all()
    mix=states.apply(pd.Series.value_counts).reindex(STATE4).fillna(0).astype(int).T; assert mix[['New','Active']].sum(axis=1).eq(am.mpu).all(); assert mix.New.eq(pf.new).all()
    life=[]
    for co,grp in coh.groupby(coh):
        obs=MONTHS[MONTHS>=co]; dead=states.loc[grp.index,obs].eq('Dead').to_numpy(); event=dead.any(axis=1); duration=np.where(event,dead.argmax(axis=1),len(obs)-1); life.append(pd.DataFrame({'cohort':co,'duration':duration,'event':event},index=grp.index))
    survival_data=pd.concat(life); km=km_table(survival_data); km_cohorts={co:km_table(g) for co,g in survival_data.groupby('cohort')}
    surv=pd.DataFrame({co:1-tab.events.cumsum().reindex(range(MONTHS[-1].ordinal-co.ordinal+1)).ffill()/sizes.loc[co] for co,tab in km_cohorts.items()}).T.reindex(index=MONTHS,columns=AGES)
    assert len(survival_data)==len(fo) and survival_data.loc[survival_data.event,'duration'].ge(DEAD_MONTHS).all()
    # LTV первой покупки; пропуск будущего периода не равен нулевой выручке.
    ce=o.join(coh,on='customer_id'); ce['age']=ce.month.astype('int64')-ce.cohort.astype('int64'); mask=cohort_mask(MONTHS); ce['order_count']=1
    monthly_amount={metric:age_matrix(ce,metric,MONTHS) for metric in ['net_revenue','contribution_margin','order_count']}
    ltv={metric:mat.cumsum(axis=1).div(sizes.replace(0,np.nan),axis=0).where(mask) for metric,mat in monthly_amount.items() if metric!='order_count'}
    ltv_summary=pd.DataFrame({'revenue_ltv':weighted_mean(ltv['net_revenue'],sizes),'margin_ltv':weighted_mean(ltv['contribution_margin'],sizes)})
    order_rate=monthly_amount['order_count'].div(sizes.replace(0,np.nan),axis=0).where(mask); margin_per_original=monthly_amount['contribution_margin'].div(sizes.replace(0,np.nan),axis=0).where(mask)
    # CAC: затраты и покупатели одной когорты входа, конверсия 30 дней.
    eligible=complete_months(CONVERSION_DAYS); converted=a30[a30.paid_30][['customer_id','entry_month']].copy(); converted.customer_id=converted.customer_id.astype(int); converted['first_month']=converted.customer_id.map(coh); assert converted.customer_id.is_unique and converted.first_month.notna().all()
    unit=pd.DataFrame({'new_buyers':converted.groupby('entry_month').size().reindex(eligible,fill_value=0),'spend':s30.groupby('entry_month').spend.sum().reindex(eligible)}); assert unit.spend.notna().all(); unit['cac']=div(unit.spend,unit.new_buyers)
    latest=pd.PeriodIndex([(m.end_time.normalize()+pd.Timedelta(days=CONVERSION_DAYS)).to_period('M') for m in eligible]); umask=cohort_mask(eligible,latest=latest)
    jo=o.merge(converted,on='customer_id',validate='many_to_one'); jo['cohort']=jo.entry_month; jo['age']=jo.month.astype('int64')-jo.first_month.astype('int64'); jo=jo[jo.age.ge(0)]
    uc_amount={metric:age_matrix(jo,metric,eligible).cumsum(axis=1).where(umask) for metric in ['net_revenue','contribution_margin']}
    uc_ltv={metric:mat.div(unit.new_buyers.replace(0,np.nan),axis=0) for metric,mat in uc_amount.items()}
    uc_ratio=uc_amount['contribution_margin'].div(unit.spend.replace(0,np.nan),axis=0).where(unit.new_buyers.gt(0),axis=0)
    n=umask.mul(unit.new_buyers,axis=0).sum(); cost=umask.mul(unit.spend,axis=0).sum(); ue=pd.DataFrame({'customers':n,'cac':div(cost,n),'revenue_ltv':div(uc_amount['net_revenue'].sum(min_count=1),n),'margin_ltv':div(uc_amount['contribution_margin'].sum(min_count=1),n),'ltv_cac':div(uc_amount['contribution_margin'].sum(min_count=1),cost)}); ue=ue[n.gt(0)]; ue['net_value']=ue.margin_ltv-ue.cac; ue['ltv_r_cac']=div(ue.revenue_ltv,ue.cac)
    assert np.allclose(ue.ltv_cac,div(ue.margin_ltv,ue.cac),equal_nan=True)
    unit['payback_month']=uc_ratio.apply(first_payback,axis=1)
    payback_age=first_payback(ue.ltv_cac)
    assert all(pd.isna(t) or (uc_ratio.loc[c,t]>=1 and uc_ratio.loc[c,:t].iloc[:-1].dropna().lt(1).all()) for c,t in unit.payback_month.items())
    cpr=div(s30.spend.sum(),a30.registered_30.sum()); cpi=div(s30[s30.entry_type.eq('app_install')].spend.sum(),len(app)); cac=div(s30.spend.sum(),a30.paid_30.sum()); acquisition_cost=pd.Series({'cpi':cpi,'cpr':cpr,'cac':cac})
    ca=pd.DataFrame({'tenure':(REPORT_END-fo.dt.normalize()).dt.days,'recency':(REPORT_END-lo.dt.normalize()).dt.days}); common=ca[ca.tenure.ge(90)]; sensitivity=pd.Series({d:common.recency.ge(d).mean() for d in [30,60,90]}); eligible60=ca[ca.tenure.ge(60)]; churn_summary=pd.Series({'eligible':len(eligible60),'inactive':eligible60.recency.ge(60).sum(),'inactive_rate':eligible60.recency.ge(60).mean()})
    return {key: value for key, value in locals().items() if key in {
        "m", "am", "eng", "wau", "f", "platform_cr", "entry", "rep", "gaps",
        "behavior", "delay", "coh", "sizes", "rates", "structure", "daily", "rolling",
        "mf", "pf", "states", "mix", "survival_data", "km", "km_cohorts", "surv",
        "ltv", "ltv_summary", "order_rate", "margin_per_original", "acquisition_cost",
        "uc_ratio", "unit", "ue", "payback_age", "sensitivity", "churn_summary"}}
```

```python
# Один слой визуализации для ноутбука и автономного HTML.
# Отклонения: robust z-score изменения относительно медианного изменения ряда.
# Не используем для накопительных кривых, воронок, когортных возрастов или перетоков.
ANOMALY_THRESHOLD=3.5

def anomaly_flags(series):
    s=series.astype(float).replace([np.inf,-np.inf],np.nan)
    delta=s.diff(); clean=delta.dropna()
    if len(clean)<8: return pd.Series(False,index=s.index)
    center=clean.median(); mad=(clean-center).abs().median()
    if mad<=1e-12: return pd.Series(False,index=s.index)
    return ((delta-center).abs()/(1.4826*mad)).gt(ANOMALY_THRESHOLD).fillna(False)

def figure():
    fig,ax=plt.subplots(figsize=(12,4.4),dpi=100,layout='none'); fig.subplots_adjust(left=.12,right=.72,top=.86,bottom=.20); return fig,ax

def legend(ax):
    handles,names=ax.get_legend_handles_labels()
    if handles: ax.legend(handles,[fill(label(n),width=25,break_long_words=False,break_on_hyphens=False) for n in names],loc='upper left',bbox_to_anchor=(1.025,1),frameon=False,fontsize=10,labelspacing=.8,borderaxespad=0)

def image_bytes(fig):
    b=BytesIO(); fig.savefig(b,format='png',dpi=105,facecolor='white'); plt.close(fig); return b.getvalue()

def plot_lines(data,title,percent=False,decimals=0,anomalies=False,xlabel='Месяц',benchmark=None,dashes=(),step=False,ylim=None,payback=None,payback_text='Первый месяц окупаемости'):
    data=data.to_frame() if isinstance(data,pd.Series) else data
    fig,ax=figure(); x=np.arange(len(data)); flagged=pd.Series(False,index=data.index)
    for i,col in enumerate(data):
        y=data[col].astype(float).to_numpy(); color=SERIES[i%len(SERIES)]; valid=np.flatnonzero(np.isfinite(y));
        if anomalies: flagged |= anomaly_flags(data[col])
        ax.plot(x,y,label=label(col),color=color,marker=None if len(x)>35 or step else 'o',linestyle='--' if col in dashes else '-',drawstyle='steps-post' if step else 'default')
        if len(valid) and len(data.columns)<=5:
            k=valid[-1]; ax.annotate(pct(y[k]) if percent else num(y[k],decimals),(x[k],y[k]),xytext=(6,0),textcoords='offset points',color=color,va='center',fontsize=9)
    for i in np.flatnonzero(flagged): ax.axvspan(i-.5,i+.5,color=PALETTE['gray'],alpha=.18,zorder=0)
    if flagged.any(): ax.plot([],[],color=PALETTE['gray'],lw=8,alpha=.3,label='Периоды с отклонениями в тренде')
    if benchmark is not None: ax.axhline(benchmark,color=PALETTE['gray'],ls='--',label='Порог покрытия CAC = 1' if '/CAC' in title else 'Порог баланса')
    if payback is not None:
        age=payback
        if pd.notna(age) and age in data.index:
            k=data.index.get_loc(age)
            v=float(data.iloc[k,0])
            ax.axvline(k,color=PALETTE['burgundy'],ls=':',lw=1.8,zorder=2,label=payback_text)
            ax.scatter(k,v,s=78,facecolors='white',edgecolors=PALETTE['burgundy'],linewidths=2,zorder=5)
            ax.annotate(f'M{int(age)} · {int(age)+1}-й мес.',(k,v),xytext=(0,16),textcoords='offset points',ha='center',va='bottom',fontsize=9,color=PALETTE['burgundy'],fontweight='bold',bbox=dict(facecolor='white',edgecolor='none',alpha=.92,pad=2))
    ticks=np.unique(np.linspace(0,max(0,len(data)-1),min(len(data),12),dtype=int)); ax.set_xticks(ticks,[str(data.index[i].date()) if isinstance(data.index[i],pd.Timestamp) else str(data.index[i].start_time.date()) if isinstance(data.index[i],pd.Period) and data.index[i].freqstr.startswith('W') else label(data.index[i]) for i in ticks],rotation=35 if len(data)>12 or isinstance(data.index,pd.PeriodIndex) else 0,ha='right' if len(data)>12 or isinstance(data.index,pd.PeriodIndex) else 'center')
    ax.set(xlabel=xlabel,xlim=(-.5,max(1,len(x)-1)+max(.65,len(x)*.13))); ax.set_title(title,loc='left'); ax.yaxis.set_major_formatter(PercentFormatter(1) if percent else FuncFormatter(lambda v,p:num(v,decimals)))
    if ylim: ax.set_ylim(*ylim)
    legend(ax); return image_bytes(fig),flagged

def plot_heat(data,title,percent=True,payback=False):
    fig,ax=plt.subplots(figsize=(12,6.5),layout="none"); fig.subplots_adjust(left=.12,right=.94,top=.88 if title else .94,bottom=.12); arr=data.to_numpy(float); finite=arr[np.isfinite(arr)]; low=0 if percent or not len(finite) else min(0,finite.min()); high=1 if percent else max(1,finite.max()) if len(finite) else 1
    im=ax.imshow(np.ma.masked_invalid(arr),aspect='auto',cmap=WARM_HEATMAP,vmin=low,vmax=high); ax.set_xticks(range(len(data.columns)),[f'M{k}' for k in data.columns]); ax.set_yticks(range(len(data.index)),data.index.astype(str)); ax.tick_params(labelsize=10); ax.grid(False); ax.set_title(title,loc='left')
    for i,j in np.ndindex(arr.shape):
        if np.isfinite(arr[i,j]): ax.text(j,i,f'{arr[i,j]:.0%}' if percent else num(arr[i,j],1 if high<10 else 0),ha='center',va='center',fontsize=10,color='white' if (arr[i,j]-low)/(high-low)>.55 else PALETTE['ink'])
    if payback:
        from matplotlib.patches import Rectangle
        for i,(_,row) in enumerate(data.iterrows()):
            age=first_payback(row)
            if pd.notna(age):
                j=data.columns.get_loc(age)
                ax.add_patch(Rectangle((j-.47,i-.47),.94,.94,fill=False,edgecolor=PALETTE['ink'],linewidth=2.1))
        fig.text(.12,.045,'Рамка — первое достижение LTV(M)/CAC ≥ 1; M0 — первый месяц.',fontsize=9,color=PALETTE['ink'])
    cb=fig.colorbar(im,ax=ax,shrink=.8); cb.ax.yaxis.set_major_formatter(PercentFormatter(1) if percent else FuncFormatter(lambda v,p:num(v,1 if high<10 else 0))); return image_bytes(fig)

def plot_bars(data,title,percent=False,horizontal=False,stacked=False,colors=None):
    fig,ax=figure(); data=data.rename(index=LABELS); data=data.rename(columns=LABELS) if isinstance(data,pd.DataFrame) else data
    if isinstance(data,pd.Series): data.plot(kind='barh' if horizontal else 'bar',ax=ax,color=PALETTE['brown'],rot=0)
    else: data.plot(kind='barh' if horizontal else 'bar',ax=ax,stacked=stacked,color=colors,rot=35)
    ax.set_title(title,loc='left'); ax.set_xlabel(''); ax.set_ylabel(''); fmt=PercentFormatter(1) if percent else FuncFormatter(lambda v,p:num(v))
    (ax.xaxis if horizontal else ax.yaxis).set_major_formatter(fmt)
    if horizontal: ax.invert_yaxis(); ax.set_xlim(0,1.16) if percent else None
    if isinstance(data,pd.Series):
        for i,v in enumerate(data):
            if pd.notna(v):
                txt=pct(v) if percent else num(v)
                if horizontal: ax.annotate(txt,(v,i),xytext=(4,0),textcoords='offset points',va='center',fontsize=9)
                else: ax.annotate(txt,(i,v),xytext=(0,5),textcoords='offset points',ha='center',fontsize=9)
    legend(ax); return image_bytes(fig)

def plot_structure(data,title,colors):
    fig,ax=figure(); v=data.dropna(); ax.stackplot(v.index,*[v[c] for c in v],colors=colors,labels=[label(c) for c in v]); ax.set(xlabel='Возраст когорты, мес.',ylim=(0,1),xticks=v.index); ax.set_title(title,loc='left'); ax.yaxis.set_major_formatter(PercentFormatter(1)); legend(ax); return image_bytes(fig)

def plot_hist(series,title,xlabel):
    fig,ax=figure(); s=series.dropna(); ax.hist(s,bins=30,color=PALETTE['brown']); ax.axvline(s.median(),color=PALETTE['burgundy'],ls='--',label='Медиана: '+num(s.median(),1)); ax.set(xlabel=xlabel,ylabel='Наблюдения'); ax.set_title(title,loc='left'); legend(ax); return image_bytes(fig)

def plot_sankey(frame):
    from matplotlib.path import Path as MP
    from matplotlib.patches import PathPatch, Rectangle
    mix=frame.apply(pd.Series.value_counts).reindex(STATE4).fillna(0).T; total=mix.sum(axis=1); scale=2.7/max(1,total.max()); coords={}; colors=dict(zip(STATE4,COL4)); fig,ax=plt.subplots(figsize=(max(12,len(MONTHS)*1.7),5),layout='none'); fig.subplots_adjust(left=.015,right=.98,bottom=.13,top=.9); ax.set(xlim=(-.2,len(MONTHS)-.08),ylim=(4.1,-.35)); ax.axis('off')
    for i,m in enumerate(MONTHS):
        y=0
        for state in STATE4:
            v=mix.loc[m,state]
            if v: coords[m,state]=(i,y,v*scale); y+=v*scale+.24
    for old,new in zip(MONTHS[:-1],MONTHS[1:]):
        mat=pd.crosstab(frame[old],frame[new]).reindex(index=STATE4,columns=STATE4,fill_value=0); out=dict.fromkeys(STATE4,0.); inc=out.copy(); assert np.array_equal(mat.sum(axis=1),mix.loc[old]); incoming=mat.sum(axis=0); incoming['New']+=mix.loc[new,'New']; assert np.array_equal(incoming,mix.loc[new])
        for a in STATE4:
            for b in STATE4:
                v=mat.loc[a,b]
                if not v: continue
                x,y,_=coords[old,a]; z,t,_=coords[new,b]; x+=.075; y+=out[a]; t+=inc[b]; h=v*scale; mid=(x+z)/2; out[a]+=h; inc[b]+=h
                verts=[(x,y),(mid,y),(mid,t),(z,t),(z,t+h),(mid,t+h),(mid,y+h),(x,y+h),(x,y)]; codes=[MP.MOVETO]+[MP.CURVE4]*3+[MP.LINETO]+[MP.CURVE4]*3+[MP.CLOSEPOLY]; ax.add_patch(PathPatch(MP(verts,codes),facecolor=colors[b],edgecolor='none',alpha=.25))
    for i,m in enumerate(MONTHS):
        ax.text(i,-.16,str(m),fontsize=9,fontweight='bold')
        for state in STATE4:
            if (m,state) not in coords: continue
            x,y,h=coords[m,state]; ax.add_patch(Rectangle((x,y),.075,h,facecolor=colors[state],edgecolor='none')); ax.text(x+.09,y+h/2,f'{num(mix.loc[m,state])} · {pct(mix.loc[m,state]/total[m])}',fontsize=7,va='center')
    ax.legend(handles=[Patch(color=colors[s],label=label(s)) for s in STATE4],loc='lower center',bbox_to_anchor=(.5,-.15),ncol=4,frameon=False); ax.set_title('Помесячные перетоки покупателей',loc='left',pad=14); return image_bytes(fig)

# Карточка = один график; наблюдения формируются на уровне раздела.
def latest(series):
    s=series.dropna(); return s.iloc[-1] if len(s) else np.nan

def change(series,percent=False):
    s=series.dropna();
    if len(s)<2: return 'Недостаточно сопоставимых наблюдений.'
    delta=s.iloc[-1]-s.iloc[-2]; change_text=f'{delta*100:+.1f} п. п.' if percent else f'{delta/s.iloc[-2]:+.1%}' if s.iloc[-2]!=0 else 'не определено (предыдущее значение 0)'
    return f'{label(series.name)}: {pct(s.iloc[-1]) if percent else num(s.iloc[-1],2 if abs(s.iloc[-1])<10 else 0)}; изменение к предыдущему периоду: {change_text}.'

def make_cards(d):
    sections = {k: [] for k in ['overview', 'engagement', 'purchase', 'retention', 'flows', 'survival', 'economics']}

    def add(section, title, image):
        sections[section].append(dict(title=title, image=image))

    def line(sec, title, data, percent=False, decimals=0, anomalies=False, xlabel='Месяц', benchmark=None, dashes=(), step=False, ylim=None, payback=None, payback_text='Первый месяц окупаемости'):
        frame = data.to_frame() if isinstance(data, pd.Series) else data
        img, _ = plot_lines(frame, title, percent, decimals, anomalies, xlabel, benchmark, dashes, step, ylim, payback, payback_text)
        add(sec, title, img)
    m = d['m']
    am = d['am']
    eng = d['eng']
    f = d['f']
    entry = d['entry']
    structure = d['structure']
    rates = d['rates']
    ue = d['ue']
    ltv = d['ltv']
    sizes = d['sizes']
    line('overview', 'Размер активной и платящей базы', am[['registered_base', 'mau', 'mpu']], anomalies=True)
    line('overview', 'Привлечение и первые покупки', am[['installs', 'web_entries', 'registrations', 'first_buyers']], anomalies=True)
    line('overview', 'Выручка и маржа, ₽', m[['revenue', 'margin']], anomalies=True)
    line('engagement', 'Количество сессий', am.sessions, anomalies=True)
    line('engagement', 'DAU / MAU', eng.DAU_MAU, percent=True, anomalies=True)
    line('engagement', 'Интенсивность использования', eng[['active_days_per_MAU', 'sessions_per_MAU']], decimals=2, anomalies=True)
    line('engagement', 'Недельная активная аудитория', d['wau'], anomalies=True, xlabel='Неделя')
    weakest = f.step_cr.idxmin()
    add('engagement', 'Воронка сессий', plot_bars(f.overall_cr, 'Конверсия от начала сессии', True, True))
    line('engagement', 'Воронка по платформам', d['platform_cr'], percent=True, xlabel='Этап')
    entry_rate = div(entry, entry.iloc[0])
    add('engagement', 'Воронка установок за 30 дней', plot_bars(entry_rate, 'Конверсия установки за 30 дней', True, True))
    for cols, title, p, dec in [(['active_customers', 'orders'], 'Покупатели и заказы', False, 0), (['aov', 'arppu'], 'Средний чек и ARPPU', False, 0), (['frequency'], 'Частота покупок', False, 2), (['margin_rate'], 'Маржинальность', True, 1)]:
        line('purchase', title, m[cols], percent=p, decimals=dec, anomalies=True)
    line('purchase', 'ARPU и ARPPU', am[['arpu', 'arppu']], anomalies=True)
    line('purchase', 'Доля покупателей среди MAU', am.payer_share, percent=True, anomalies=True)
    line('purchase', 'Повторная покупка: зрелая база', d['rep'].rate, percent=True, xlabel='Дни после первого заказа')
    for s, title, x in [(d['gaps'], 'Интервалы между заказами', 'Дни'), (d['behavior'].orders, 'Заказы за полугодие', 'Количество заказов')]:
        add('purchase', title, plot_hist(s, title, x))
    for state in STATE3:
        mat = rates[state]
        add('retention', label(state), plot_heat(mat, label(state)))
    add('retention', 'Структура когорт', plot_structure(structure, 'Структура когорт: 100%', COL3))
    line('retention', 'Размер отточной базы', d['daily'].churned_customers, anomalies=True, xlabel='Дата')
    line('retention', 'Доля текущей отточной базы', d['daily'].inactive_rate.rename('Churn'), percent=True, anomalies=True, xlabel='Дата')
    line('retention', 'Невозврат: окно 180 / 60 дней', d['rolling'].churn_rate, percent=True, anomalies=True, xlabel='Дата формирования базы')
    add('retention', 'Чувствительность к порогу', plot_bars(d['sensitivity'], 'Неактивность: 30 / 60 / 90 дней', True))
    for name, flow in [('MAU', d['mf']), ('MPU', d['pf'])]:
        b = flow[['new', 'reactivated', 'churned']].copy()
        b.churned *= -1
        last = flow.iloc[-1]
        add('flows', f'{name}: приток и потери', plot_bars(b, f'{name}: приток и потери', stacked=True, colors=[FLOW_COLORS[c] for c in b]))
        line('flows', f'{name}: Quick Ratio', flow.quick_ratio, decimals=2, benchmark=1)
    mix = d['mix']
    add('flows', 'Жизненный цикл: Sankey', plot_sankey(d['states']))
    add('survival', 'Когортная выживаемость', plot_heat(d['surv'], ''))
    km = d['km']
    line('survival', 'Каплан — Майер: все когорты', km.survival, percent=True, xlabel='Возраст когорты, мес.', step=True, ylim=(0, 1.05))
    curves = d['surv'].T
    line('survival', 'Кривые отдельных когорт', curves, percent=True, xlabel='Возраст когорты, мес.', step=True, ylim=(0, 1.05))
    for key, title in [('order_rate', 'Заказы на исходного покупателя'), ('margin_per_original', 'Маржа на исходного покупателя')]:
        add('economics', title, plot_heat(d[key], title, False))
    for metric, title in [('net_revenue', 'Накопленный LTV(R)'), ('contribution_margin', 'Накопленный LTV(M)')]:
        add('economics', title, plot_heat(ltv[metric], title, False))
    line('economics', 'Средневзвешенный накопленный LTV', d['ltv_summary'], xlabel='Возраст когорты, мес.')
    add('economics', 'Стоимость привлечения', plot_bars(d['acquisition_cost'], 'CPI, CPR, CAC, ₽'))
    add('economics', 'Когортный LTV(M)/CAC', plot_heat(d['uc_ratio'], 'Когортный LTV(M)/CAC', False, payback=True))
    line('economics', 'LTV(M) и CAC', ue[['margin_ltv', 'cac']], xlabel='Возраст с первой покупки, мес.', dashes=('cac',), payback=d['payback_age'])
    line('economics', 'Динамика LTV(M)/CAC', ue.ltv_cac, decimals=2, xlabel='Возраст с первой покупки, мес.', benchmark=1, payback=d['payback_age'])
    line('economics', 'Динамика LTV(R)/CAC', ue.ltv_r_cac, decimals=2, xlabel='Возраст с первой покупки, мес.', benchmark=1, payback=first_payback(ue.ltv_r_cac), payback_text='Первое покрытие CAC выручкой')
    return sections
```

### 5.2 / Движение клиентской базы · Customer Base Dynamics

New — активен впервые за наблюдаемую историю; Retained — активен в обоих месяцах; Reactivated — активен сейчас, пропустил предыдущий месяц, но был активен раньше; Churned — активен в прошлом месяце, неактивен сейчас.

### 5.4 / Когортная выживаемость · Cohort Survival Analysis

**Бизнес-вопросы:** как долго после первой покупки клиенты не достигают глубокой неактивности? На каком возрасте когорты происходит наиболее заметное сокращение доли еще не достигших этого состояния? Отличаются ли когорты по длительности жизни?

**Зачем бизнесу:** сравнивать устойчивость когорт на одинаковом возрасте, выделять периоды для дальнейшего исследования удержания. В отличие от Retention, выживаемость здесь отражает время до первого события, а не факт покупки в конкретном месяце.

**Определение события:** первое попадание клиента в Dead (три подряд календарных месяца без заказа после первой покупки), согласно разделу 5.3. Возраст измеряем в месяцах с первой покупки (M0). После первого события клиент больше не считается выжившим для этой метрики, даже если позднее вернулся. Это не утверждение о необратимой потере клиента.


**Метод и визуализация:** оценка Каплана–Майера на дискретной шкале месяцев: $S(t)=\prod_{j\le t}(1-d_j/n_j)$, где $d_j$ — число первых событий на возрасте $j$, $n_j$ — число клиентов под риском непосредственно перед событиями на этом возрасте. Цензурирование на конце месяца учитывается после событий этого месяца. Тепловая карта позволяет сравнивать когорты на одинаковом возрасте; ступенчатая кривая показывает накопленную долю покупателей без первого события глубокой неактивности. Общая кривая объединяет доступные наблюдения всех когорт, ее состав меняется с возрастом; объединение требует осторожности при различиях между когортами. Она не является прогнозом будущей жизни клиента.

#### 6.2.2. LTV(M)/CAC и Payback Period

**Бизнес-вопросы:** Покрывает ли накопленная маржа расходы на привлечение и в каком месяце?

**Payback Period** — первый наблюдаемый возраст, при котором накопленный LTV(M) ≥ CAC (LTV(M)/CAC ≥ 1). Горизонталь на уровне 1 показывает порог, вертикальная линия и маркер — первый месяц достижения. В когортной тепловой карте рамкой выделяется первое достижение порога для каждой когорты; в таблице срок указан отдельным столбцом. M0 — первый календарный месяц покупки, M3 — четвертый, M4 — пятый. Это не точный день окупаемости и не гарантия устойчивой окупаемости: при отрицательной марже отношение может снова снизиться.

Состав наблюдаемых когорт в средневзвешенной кривой меняется с возрастом, поэтому ее пересечение — агрегированный ориентир, а индивидуальные сроки когорт приведены отдельно. Расходы в синтетическом сценарии повышены для учебной демонстрации, не для имитации рыночного бенчмарка.
